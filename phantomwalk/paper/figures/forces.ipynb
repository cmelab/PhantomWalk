```python
import matplotlib
import matplotlib_inline
import matplotlib.pyplot as plt
%matplotlib inline
matplotlib.style.use("default")
matplotlib_inline.backend_inline.set_matplotlib_formats("svg")
import numpy as np
```

```python
A = 250000
gamma = 250000
r_cut = 1.01
x = np.linspace(0, r_cut, 100)
fig, ax = plt.subplots()
ax.plot(x, A*(1-(x/r_cut)), label=r"$F_C=A\left(1-\frac{r_{ij}}{r_{cut}}\right)$")
ax.plot(x, gamma*(1-(x/r_cut))**2, linestyle="dotted", label=r"$F_D=\gamma\left(1-\frac{r_{ij}}{r_{cut}}\right)^2$")
ax.set_title(fr"$\gamma = {gamma}$, $A = {A}$, $r_{{cut}} = {r_cut}$",fontsize=12)
plt.suptitle("DPD Conservative and Dissipative Forces", fontsize=12)
ax.legend(fontsize=12)
plt.xlabel(r"$r_{ij}$", fontsize=12)
plt.savefig("forces.svg")
```
