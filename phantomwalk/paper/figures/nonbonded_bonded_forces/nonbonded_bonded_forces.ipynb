```python
import matplotlib
import matplotlib_inline
import matplotlib.pyplot as plt
%matplotlib inline
matplotlib.style.use("default")
matplotlib_inline.backend_inline.set_matplotlib_formats("svg")
import numpy as np
```

```python
A = 250000
k_b = 250000
r_cut = 1.01
x = np.linspace(0, r_cut, 100)
fig, ax = plt.subplots()
ax.plot(x, A*(1-(x/r_cut)), label=r"$F_C=A\left(1-\frac{r_{ij}}{r_{cut}}\right)$")
ax.plot(x, 0.5*k_b*(1-x)**2, linestyle="dotted", label=r"$F_B=\frac{1}{2}k_b(1-r_{ij})^2$")
ax.set_title(fr"$k_b = {k_b}$, $A = {A}$, $r_{{cut}} = {r_cut}$",fontsize=12)
plt.suptitle(r"DPD Non-Bonded ($F_C$) and Bonded ($F_B$) Forces", fontsize=12)
ax.legend(fontsize=12)
plt.xlabel(r"$r_{ij}$", fontsize=12)
plt.savefig("nonbonded_bonded_forces.svg")
```
